```python
%%writefile hola3.cpp
/*3. Code the Middle-school procedure algorithm from the textbook (You will need to code the
Sieve of Eratosthenes algorithm too).*/

#include <iostream>
#include <vector>
#include <algorithm>

using namespace std;

vector<int> Sieve(int n) {
    vector<int> A(n + 1);
    for (int p = 2; p <= n; ++p) {
        A[p] = p;
    }
    for (int p = 2; p * p <= n; ++p) {
        if (A[p] != 0) {
            int j = p * p;
            while (j <= n) {
                A[j] = 0;
                j = j + p;
            }
        }
    }
    vector<int> L;
    for (int p = 2; p <= n; ++p) {
        if (A[p] != 0) {
            L.push_back(A[p]);
        }
    }
    return L;
}

vector<int> Factors(int num, const vector<int>& primes) {
    vector<int> f;
    for (int prime : primes) {
        if (prime * prime > num && num == 1) break;
        while (num % prime == 0) {
            f.push_back(prime);
            num /= prime;
        }
    }
    if (num > 1) {
        f.push_back(num);
    }
    return f;
}

int Gcd(int m, int n) {
    int limit = max(m, n);

    vector<int> primes = Sieve(limit);
    vector<int> fm = Factors(m, primes);
    vector<int> fn = Factors(n, primes);

    int res = 1;
    size_t i = 0, j = 0;
    while (i < fm.size() && j < fn.size()) {
        if (fm[i] == fn[j]) {
            res *= fm[i];
            i++;
            j++;
        } else if (fm[i] < fn[j]) {
            i++;
        } else {
            j++;
        }
    }
    return res;
}

int main() {
    int m, n;
    cin >> m >> n;
    cout << Gcd(m, n) << endl;
    return 0;
}


```

```python
%%writefile hola2.cpp
/*5.  Create a class that encapsulates the stack behavior describe in point 4 (Ex: MyStack*/

#include <cstdlib>

typedef struct info {
    int key;
    int value;
} Info;

class MyStack {
private:
    Info *data;
    int size;
    int n;
    int top;

public:
    MyStack(int stack_size) {
        data = (Info *)malloc(sizeof(Info) * stack_size);
        size = stack_size;
        n = 0;
        top = 0;
    }

    bool stack_is_empty() {
        return (n == 0);
    }

    bool stack_is_full() {
        return (n == size);
    }

    void push(Info item) {
        if (!stack_is_full()) {
            data[top] = item;
            top++;
            n++;
        }
    }

    Info pop() {
        if (!stack_is_empty()) {
            top--;
            n--;
            return data[top];
        }
        Info empty_item = {0, 0};
        return empty_item;
    }
};



```

```python
%%writefile hola2.cpp
/*7. find the way C++ uses to queue and dequeue elements to and from a list to behave like a
queue. Create an example that showcases the idea. */

#include <iostream>
#include <stdexcept>

using namespace std;

template <typename T>
class MyQueue {
private:
    T *data_array;
    unsigned int array_size;
    unsigned int head;
    unsigned int tail;
    unsigned int n;

public:
    MyQueue(int n_size = 5) {
        data_array = new T[n_size];
        array_size = n_size;
        head = 0;
        tail = 0;
        n = 0;
    }

    ~MyQueue() {
        delete[] data_array;
    }

    inline bool isEmpty() {
        return n == 0;
    }

    inline bool isFull() {
        return n == array_size;
    }

    void enqueue(T element) {
        if (!isFull()) {
            data_array[tail] = element;
            tail = (tail + 1) % array_size;
            n++;
        }
    }

    T dequeue() {
        if (!isEmpty()) {
            T element = data_array[head];
            head = (head + 1) % array_size;
            n--;
            return element;
        }
        throw runtime_error("Queue is empty");
    }
};
```

```python
%%writefile hola2.cpp
/*9. find the way C++ uses to queue and dequeue elements to and from a list to behave like a
queue. Create an example that showcases the idea. */


#include <iostream>

using namespace std;

template <typename T>
class Node {
public:
    T data;
    Node *left;
    Node *right;
    Node *parent;

    Node(T val) {
        data = val;
        left = nullptr;
        right = nullptr;
        parent = nullptr;
    }
};

template <typename T>
class BinaryTree {
private:
    void postorderRecursiveDelete(Node<T> *current) {
        if (current != nullptr) {
            postorderRecursiveDelete(current->left);
            postorderRecursiveDelete(current->right);
            cout << current->data << endl;
            delete current;
        }
    }

    Node<T>* insertRecursive(Node<T> *current, T val, Node<T> *parentPtr) {
        if (current == nullptr) {
            Node<T> *newNode = new Node<T>(val);
            newNode->parent = parentPtr;
            return newNode;
        }

        if (val < current->data) {
            current->left = insertRecursive(current->left, val, current);
        } else if (val > current->data) {
            current->right = insertRecursive(current->right, val, current);
        }

        return current;
    }

    void inorderRecursive(Node<T> *current) {
        if (current != nullptr) {
            inorderRecursive(current->left);
            cout << current->data << " ";
            inorderRecursive(current->right);
        }
    }

public:
    Node<T> *root;

    BinaryTree() {
        root = nullptr;
    }

    ~BinaryTree() {
        postorderRecursiveDelete(root);
    }

    void insert(T val) {
        root = insertRecursive(root, val, nullptr);
    }

    void printInorder() {
        inorderRecursive(root);
        cout << endl;
    }
};

int main() {
    BinaryTree<int> tree;

    tree.insert(50);
    tree.insert(30);
    tree.insert(70);
    tree.insert(20);
    tree.insert(40);

    tree.printInorder();

    return 0;
}
```
