```python
%%writefile hola.cpp
/* 1. Implement the forward elimination algorithm for the Gaussian elimination. */

#include <iostream>
#include <vector>
#include <iomanip>

void forwardElimination(std::vector<std::vector<double>>& A, std::vector<double>& b) {
    int n = A.size();
    for (int k = 0; k < n - 1; ++k) {
        for (int i = k + 1; i < n; ++i) {
            double factor = A[i][k] / A[k][k];
            for (int j = k; j < n; ++j) {
                A[i][j] -= factor * A[k][j];
            }
            b[i] -= factor * b[k];
        }
    }
}

int main() {
    std::cout << std::fixed << std::setprecision(4);

    std::vector<std::vector<double>> A = {
        {2, 1, -1},
        {-3, -1, 2},
        {-2, 1, 2}
    };
    std::vector<double> b = {8, -11, -3};

    std::cout << "Original Matrix [A | b]:\n";
    for (size_t i = 0; i < A.size(); ++i) {
        for (size_t j = 0; j < A[i].size(); ++j) {
            std::cout << A[i][j] << "\t";
        }
        std::cout << "| " << b[i] << "\n";
    }

    forwardElimination(A, b);

    std::cout << "\nForward Elimination:\n";
    for (size_t i = 0; i < A.size(); ++i) {
        for (size_t j = 0; j < A[i].size(); ++j) {
            std::cout << A[i][j] << "\t";
        }
        std::cout << "| " << b[i] << "\n";
    }

    return 0;
}
```

```python
%%writefile hola2.cpp
/* 2. Design the backwards substitution algorithm and show that its running time is in O(n^2) (The probe of the complexity is in the pdf).*/

#include <iostream>
#include <vector>
#include <iomanip>

std::vector<double> backwardSubstitution(const std::vector<std::vector<double>>& A, const std::vector<double>& b) {
    int n = A.size();
    std::vector<double> x(n);
    for (int i = n - 1; i >= 0; --i) {
        double sum = 0.0;
        for (int j = i + 1; j < n; ++j) {
            sum += A[i][j] * x[j];
        }
        x[i] = (b[i] - sum) / A[i][i];
    }
    return x;
}

int main() {
    std::cout << std::fixed << std::setprecision(4);

    // Upper triangular matrix from Exercise 1 result
    std::vector<std::vector<double>> A = {
        {2.0, 1.0, -1.0},
        {0.0, 0.5, 0.5},
        {0.0, 0.0, 3.0}
    };
    std::vector<double> b = {8.0, 1.0, 3.0};

    std::vector<double> x = backwardSubstitution(A, b);

    std::cout << "Solution Vector x:\n";
    for (size_t i = 0; i < x.size(); ++i) {
        std::cout << "  x[" << i << "] = " << x[i] << "\n";
    }

    return 0;
}
```

```python
%%writefile hola3.cpp
/* 3. Implement an algorithm to transform a random array into a heap. */

#include <iostream>
#include <vector>
#include <random>

void heapify(std::vector<int>& arr, int n, int i) {
    int largest = i;
    int left = 2 * i + 1;
    int right = 2 * i + 2;

    if (left < n && arr[left] > arr[largest]) {
        largest = left;
    }
    if (right < n && arr[right] > arr[largest]) {
        largest = right;
    }

    if (largest != i) {
        std::swap(arr[i], arr[largest]);
        heapify(arr, n, largest);
    }
}

void buildHeap(std::vector<int>& arr) {
    int n = arr.size();
    for (int i = n / 2 - 1; i >= 0; --i) {
        heapify(arr, n, i);
    }
}

int main() {
    std::random_device rd;
    std::mt19937 gen(12345);
    std::uniform_int_distribution<int> dis(1, 100);

    int n = 10;
    std::vector<int> arr(n);
    for (int i = 0; i < n; ++i) {
        arr[i] = dis(gen);
    }

    std::cout << "Random Array:\n";
    for (int val : arr) {
        std::cout << val << " ";
    }
    std::cout << "\n\n";

    buildHeap(arr);

    std::cout << "Heap Array:\n";
    for (int val : arr) {
        std::cout << val << " ";
    }
    std::cout << "\n";

    return 0;
}
```

```python
%%writefile hola4.cpp
/* 4. Implement an algorithm that can solve a system of equations using LU decomposition.
   LU decomposition is based on Gaussian Elimination. */

#include <iostream>
#include <vector>
#include <iomanip>

void luDecomposition(const std::vector<std::vector<double>>& A,
                     std::vector<std::vector<double>>& L,
                     std::vector<std::vector<double>>& U) {
    int n = A.size();
    L = std::vector<std::vector<double>>(n, std::vector<double>(n, 0.0));
    U = A;

    for (int i = 0; i < n; ++i) {
        L[i][i] = 1.0;
    }

    for (int k = 0; k < n - 1; ++k) {
        for (int i = k + 1; i < n; ++i) {
            double factor = U[i][k] / U[k][k];
            L[i][k] = factor;
            for (int j = k; j < n; ++j) {
                U[i][j] -= factor * U[k][j];
            }
        }
    }
}

std::vector<double> forwardSubstitution(const std::vector<std::vector<double>>& L,
                                         const std::vector<double>& b) {
    int n = L.size();
    std::vector<double> y(n);
    for (int i = 0; i < n; ++i) {
        double sum = 0.0;
        for (int j = 0; j < i; ++j) {
            sum += L[i][j] * y[j];
        }
        y[i] = (b[i] - sum) / L[i][i];
    }
    return y;
}

std::vector<double> backwardSubstitution(const std::vector<std::vector<double>>& U,
                                          const std::vector<double>& y) {
    int n = U.size();
    std::vector<double> x(n);
    for (int i = n - 1; i >= 0; --i) {
        double sum = 0.0;
        for (int j = i + 1; j < n; ++j) {
            sum += U[i][j] * x[j];
        }
        x[i] = (y[i] - sum) / U[i][i];
    }
    return x;
}

int main() {
    std::cout << std::fixed << std::setprecision(4);

    std::vector<std::vector<double>> A = {
        {2, 1, -1},
        {-3, -1, 2},
        {-2, 1, 2}
    };
    std::vector<double> b = {8, -11, -3};

    std::vector<std::vector<double>> L, U;
    luDecomposition(A, L, U);

    std::cout << "Matrix L :\n";
    for (const auto& row : L) {
        for (double val : row) std::cout << val << "\t";
        std::cout << "\n";
    }

    std::cout << "\nMatrix U :\n";
    for (const auto& row : U) {
        for (double val : row) std::cout << val << "\t";
        std::cout << "\n";
    }

    std::vector<double> y = forwardSubstitution(L, b);
    std::vector<double> x = backwardSubstitution(U, y);

    std::cout << "\nSolution Vector x:\n";
    for (size_t i = 0; i < x.size(); ++i) {
        std::cout << "  x[" << i << "] = " << x[i] << "\n";
    }

    return 0;
}
```
